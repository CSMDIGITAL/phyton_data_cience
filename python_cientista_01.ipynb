```python
# primeiro parenteses
# dos internos para os externos
# não há colchetes ou chaves nas expressões python
# se for necessário aninham-se parenteses
# exponenciação
# divisão
# multiplicação
# adição
# subtração
calculo_exemplo = 10 + 5 * 2
print(calculo_exemplo)

outro_calculo = (10 + 5) * 2
print(outro_calculo)

mais_outro_calculo = 10 + 5 * 2 ** 2
print(mais_outro_calculo)

exemplo_parententes_aninhado = (10 + 5) * (2 ** 2)
print(exemplo_parententes_aninhado)

exemplo_completao = 10 + 5 * 2 ** 2 / 2
print(exemplo_completao)
```
